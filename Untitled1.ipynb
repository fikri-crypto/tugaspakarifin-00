```python
import pandas as pd

```

```python
df = pd.read_csv("data_kantin - data_kantin.csv")
```

```python
display(df.head())
```

```python
print("Informasi Dataset:")
df.info()
```

```python
print("Statistik Deskriptif:")
display(df.describe())
```

```python
print("Ukuran dataset:", df.shape)
```

```python
df["tanggal"] = pd.to_datetime(df["tanggal"])
df["harga"] = pd.to_numeric(df["harga"], errors="coerce")
df["terjual"] = pd.to_numeric(df["terjual"], errors="coerce")
```

```python
print("Missing value sebelum dibersihkan:")
print(df.isnull().sum())
```

```python
df["menu"] = df["menu"].fillna("Tidak Diketahui")
```

```python
df["terjual"] = df.groupby("menu")["terjual"].transform(
    lambda x: x.fillna(x.median())
)
```

```python
df["terjual"] = df["terjual"].fillna(df["terjual"].median())
```

```python
df = df.drop_duplicates()
```

```python
print("Missing value setelah dibersihkan:")
print(df.isnull().sum())
```

```python
print("Jumlah data setelah cleaning:", df.shape)
```

```python
df["pendapatan"] = df["harga"] * df["terjual"]
```

```python
data_makanan = df[df["kategori"] == "Makanan"]
```

```python
print("Data kategori Makanan:")
display(data_makanan)
```

```python
data_urut = df.sort_values(
    by="pendapatan",
    ascending=False
)
```

```python
print("Data berdasarkan pendapatan terbesar:")
display(data_urut.head(10))
```

```python
ringkasan_menu = df.groupby("menu").agg(
    total_terjual=("terjual", "sum"),
    total_pendapatan=("pendapatan", "sum")
).sort_values(
    by="total_terjual",
    ascending=False
)
```

```python
print("Ringkasan berdasarkan menu:")
display(ringkasan_menu)
```

```python
ringkasan_kategori = df.groupby("kategori").agg(
    total_terjual=("terjual", "sum"),
    total_pendapatan=("pendapatan", "sum")
).sort_values(
    by="total_pendapatan",
    ascending=False
)
```

```python
print("Ringkasan berdasarkan kategori:")
display(ringkasan_kategori)
```
